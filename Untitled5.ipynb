```python
import pandas as pd

print("Downloading dataset from the new OWID Catalog... (this might take a minute)")
# The updated, active URL for the complete OWID COVID-19 dataset
url = 'https://catalog.ourworldindata.org/garden/covid/latest/compact/compact.csv'
df = pd.read_csv(url)

print(f"Success! Loaded {len(df):,} rows and {len(df.columns)} columns.")
```

```python
import numpy as np

# 1. Filter out aggregate regions (like 'World', 'Europe') to only keep real countries
# OWID uses iso_codes starting with 'OWID_' for regional aggregates
# The 'iso_code' column was not found. Filtering by known aggregate names in 'country' column instead.
aggregate_locations = ['World', 'Europe', 'Asia', 'North America', 'South America',
                       'Africa', 'Oceania', 'European Union', 'High income',
                       'Low income', 'Upper middle income', 'Lower middle income',
                       'International']
df = df[~df['country'].isin(aggregate_locations)].copy()

# 2. Ensure proper datetime typing and chronological sorting
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values(by=['country', 'date']).reset_index(drop=True)

# 3. Handle Missing Values (Documented Assumptions for the Project Brief)
# Assumption 1: Missing daily counts (cases, deaths) mean 0 reported for that day.
count_cols = ['new_cases', 'new_deaths', 'new_vaccinations']
for col in count_cols:
    if col in df.columns:
        df[col] = df[col].fillna(0)
        df.loc[df[col] < 0, col] = 0 # Clean negative reporting corrections

# Assumption 2: Missing cumulative totals mean no change from the previous known day.
cum_cols = ['total_cases', 'total_deaths', 'total_vaccinations', 'people_fully_vaccinated']
for col in cum_cols:
    if col in df.columns:
        df[col] = df.groupby('country')[col].ffill().fillna(0)

# 4. Feature Engineering: Required Derived Metrics
# A. 7-Day Rolling Averages
df['new_cases_7day_avg'] = df.groupby('country')['new_cases'].rolling(window=7, min_periods=1).mean().reset_index(drop=True)
df['new_deaths_7day_avg'] = df.groupby('country')['new_deaths'].rolling(window=7, min_periods=1).mean().reset_index(drop=True)

# B. Case Fatality Rate (CFR %)
df['case_fatality_rate_pct'] = np.where(df['total_cases'] > 0, (df['total_deaths'] / df['total_cases']) * 100, 0)

# C. Doubling Time (Calculated over a 7-day growth window)
cases_7_days_ago = df.groupby('country')['total_cases'].shift(7)
growth_ratio = df['total_cases'] / cases_7_days_ago

# Create a safe argument for np.log to prevent RuntimeWarning when growth_ratio <= 1
log_arg = growth_ratio.astype(float)
log_arg[log_arg <= 1] = np.nan # Set to NaN if growth_ratio is 1 or less

# Calculate log of the growth ratio, which will now avoid log(1) and log(0)
log_growth_ratio_safe = np.log(log_arg)

# Using 7 * ln(2) / ln(Total Cases Today / Total Cases 7 Days Ago)
df['doubling_time_days'] = np.where(
    (growth_ratio > 1) & (cases_7_days_ago > 0), # Only calculate if there's actual growth and cases 7 days ago
    7 * np.log(2) / log_growth_ratio_safe,
    np.nan
)

# D. Normalize by Population (Per Million metrics for fair country comparisons)
if 'population' in df.columns:
    df['new_cases_per_million'] = (df['new_cases'] / df['population']) * 1_000_000
    df['total_deaths_per_million'] = (df['total_deaths'] / df['population']) * 1_000_000
    df['fully_vaccinated_pct'] = np.where(df['population'] > 0, (df['people_fully_vaccinated'] / df['population']) * 100, 0)

# 5. Export the cleaned deliverable
output_filename = 'cleaned_covid_data.csv'
df.to_csv(output_filename, index=False)
print(f"✅ Cleaned dataset ready: {output_filename}")
print(f"✅ Final shape: {df.shape[0]:,} rows and {df.shape[1]} columns.")
```
